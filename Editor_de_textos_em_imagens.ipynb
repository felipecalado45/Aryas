### Etapa 2: Código base para Detecção de Texto (OCR)
O código abaixo mostra como carregar uma imagem, extrair os textos com suas posições (caixas delimitadoras) e exibir o resultado. (Você precisará fazer o upload de uma imagem para testar).

```python
import easyocr
import cv2
import matplotlib.pyplot as plt

# Inicializa o leitor OCR em português e inglês
reader = easyocr.Reader(['pt', 'en'], gpu=True)

def extrair_texto(caminho_imagem):
    # Lê a imagem e executa o OCR
    resultados = reader.readtext(caminho_imagem)

    # Carrega a imagem para visualização
    img = cv2.imread(caminho_imagem)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    print("Textos extraídos:")
    for (bbox, texto, confianca) in resultados:
        print(f"- {texto} (Confiança: {confianca:.2f})")

        # Desenha a caixa delimitadora
        (top_left, top_right, bottom_right, bottom_left) = bbox
        top_left = (int(top_left[0]), int(top_left[1]))
        bottom_right = (int(bottom_right[0]), int(bottom_right[1]))
        cv2.rectangle(img, top_left, bottom_right, (0, 255, 0), 2)

    plt.figure(figsize=(10, 10))
    plt.imshow(img)
    plt.axis('off')
    plt.show()

# Exemplo de uso (substitua 'sua_imagem.jpg' pelo caminho de uma foto sua)
# extrair_texto('sua_imagem.jpg')
```
